```python
# while loop

# while condition:
#     # body

# Every while loop needs three things:
# 1. A starting value, set BEFORE the loop
# 2. A condition, checked before EACH pass
# 3. An update INSIDE the loop that moves toward making the condition False

balance = 0
months = 0
while balance < 1000:
    balance += 150
    months += 1
print(months, balance)

month = 0, 1, 2, 3, 4, 5, 6, 7
balance = 0, 150, 300, 450, 600, 750, 900, 1050

# Use for    -> when you have a sequence or a known number of repetitions
# Use while  -> when you repeat until a condition changes and
#            do not know in advance how many passes it will take
```

```python
# break      → leave the loop entirely, right now
# continue   → skip the rest of THIS pass, go on to the next pass

for n in range(1, 6):
    if n == 4:
        break # break the loop and rest of the iterations wont't execute.
    print(n)

print('========')

for n in range(1, 6):
    if n == 3:
        continue # pass the current iteration.
    print(n)
```

```python
# Scope of the argument

# A variable defined inside a function can't be accessed outside the function.

def func1(price, qty):
  total = price * qty
  return total

def func2(price, qty, discount):
  total = price * qty * (1-discount)
  return total

# print(func(100, 5))
# print(total)
# func(100, 5)

print(func1(100, 5))

print(func2(100, 5, 0.2))

# Each function get's its own scope or workspace.
# When a function is called, python creates a private workspace for that call.

name = 'Krunal'

name = 'Ved Prakash'

print(name)
```

```python
# Lists and Strings Handling in Python

# List - Holds multiple values in order.
# String - Text

# numbers = []
# append, extend, sorted, ......


product_codes = ["KRT-0412-RD", "Masai", "India", "Machine Learning",
        "JNS-0331-BL", "jns-0332-bk", "SAR-0905-RD", "Dubai",
        "SHO-0620-WH", "SHO-0621-BK"]

# indexing => 0 to N-1
#
print(product_codes)
print(len(product_codes)) # number of elements in the list or size.

# elements in the list are indexed from 0.
print(product_codes[1])

print(product_codes[-10])
print(product_codes[-1]) # last element of the list.
print(product_codes[-2])

# print(product_codes[-11])

print(product_codes[9] == product_codes[-1])

# slice function

# Start at position 2 and stop before position 5. Position 5 is not included.
print(product_codes[2:5]) # returns a new list from start (inclusive) to end (exclusive)


# Leaving out start means start at the front. This gives the first three.
print(product_codes[:3]) # start index - 0

print(product_codes[2:8:2]) # start, end, step


print(product_codes[7:]) # take the list from 7th index to the end.
print(product_codes[-3:]) # start = -3, end is not given, that means go till the end of the list.



print(product_codes[::3]) # start = ?, end = ?, step = 3

print(product_codes[::-1]) # gives the list in reverse order.

print(product_codes[5:20]) # start from index 5 and go till the end of the list.
print(product_codes[5:])
```
