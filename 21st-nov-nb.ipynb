```python
# while loop

# while condition:
#     # body

# Every while loop needs three things:
# 1. A starting value, set BEFORE the loop
# 2. A condition, checked before EACH pass
# 3. An update INSIDE the loop that moves toward making the condition False

balance = 0
months = 0
while balance < 1000:
    balance += 150
    months += 1
print(months, balance)

month = 0, 1, 2, 3, 4, 5, 6, 7
balance = 0, 150, 300, 450, 600, 750, 900, 1050

# Use for    -> when you have a sequence or a known number of repetitions
# Use while  -> when you repeat until a condition changes and
#            do not know in advance how many passes it will take
```

```python
# break      → leave the loop entirely, right now
# continue   → skip the rest of THIS pass, go on to the next pass

for n in range(1, 6):
    if n == 4:
        break # break the loop and rest of the iterations wont't execute.
    print(n)

print('========')

for n in range(1, 6):
    if n == 3:
        continue # pass the current iteration.
    print(n)
```

```python
# Functions & Reusable Code.

# DRY - Don't Repeat Yourself.

# Insead of repeating the same code multiple times at different places in the codebase, we can write the code inside a function and use that function at all the places where it is required.

# def function_name(parameter_one, parameter_two):
  # function body.

# - def : start the definition of the function
# - parameters are the input passed to the function.
#
# def totalPrice()

def total_price_return(unit_price, quantity):
  return unit_price * quantity

def total_price_print(unit_price, quantity):
  print(unit_price * quantity)

result1 = total_price_return(1000, 2)
result2 = total_price_print(1000, 2)

# result += 10

print(result1)
print(result2)

# print(total_price(1000, 8))

# print(total_price(540.5, 10))

# Print vs Return - print displays a value on the screen and then the value is gone,
# Return hands the value back to the function that can be used later.

```

```python
# Scope of the argument

# A variable defined inside a function can't be accessed outside the function.

def func1(price, qty):
  total = price * qty
  return total

def func2(price, qty, discount):
  total = price * qty * (1-discount)
  return total

# print(func(100, 5))
# print(total)
# func(100, 5)

print(func1(100, 5))

print(func2(100, 5, 0.2))

# Each function get's its own scope or workspace.
# When a function is called, python creates a private workspace for that call.

name = 'Krunal'

name = 'Ved Prakash'

print(name)
```

```python
total = 1000

def add_fee(price):
  total = price + 10
  return total


# print(price)
print(add_fee(90))
print(total)

# Assigning to total inside the function created a brand new local variable, separate from the outer total.
# They share a name but they are different variables in different workspaces. The outer one still holds 1000

# A variable created inside a function is a local variable and exists only in that call's workspace

# Each call creates a fresh workspace, which is discarded when the function returns; only the returned value comes out

# Parameters are local variables too

# A local variable can share a name with an outer variable, but they are different variables, and assigning inside does not change the outer one

# To get a value out of a function, return it; to give a function a value, pass it in as an argument

```

```python
# Class - Template or Blueprint. Defined once, a class describes what data, what actions.


# __ (Double underscore) =>  dunder
class Student:
  # Constructor
  def __init__(self, input_name, input_email, input_age, input_phone):
    self.name = input_name
    self.email = input_email
    self.age = input_age
    self.phone = input_phone


# Object / Instance  - Real entity created from the template.
# Any number of objects or instances can be created from a class, and each object is going to have it's own data.

student1 = Student('Vaibhav', 'vaibhav@gmail.com', 20, '1234567')


student2 = Student('Sanjeev', 'sanjeev@gmail.com', 21, '29374789')

print(student1.name)
print(student2.name)

print(student1.age)
print(student2.age)
```
