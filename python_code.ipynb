```python
# while loop

# while condition:
#     # body

# Every while loop needs three things:
# 1. A starting value, set BEFORE the loop
# 2. A condition, checked before EACH pass
# 3. An update INSIDE the loop that moves toward making the condition False

balance = 0
months = 0
while balance < 1000:
    balance += 150
    months += 1
print(months, balance)

month = 0, 1, 2, 3, 4, 5, 6, 7
balance = 0, 150, 300, 450, 600, 750, 900, 1050

# Use for    -> when you have a sequence or a known number of repetitions
# Use while  -> when you repeat until a condition changes and
#            do not know in advance how many passes it will take
```

```python
# break      → leave the loop entirely, right now
# continue   → skip the rest of THIS pass, go on to the next pass

for n in range(1, 6):
    if n == 4:
        break # break the loop and rest of the iterations wont't execute.
    print(n)

print('========')

for n in range(1, 6):
    if n == 3:
        continue # pass the current iteration.
    print(n)
```

```python
students = {
    "Rahul": 85,
    "Amit": 32,
    "Priya": 92,
    "Neha": 28,
    "Rohan": 76,
    "Sneha": 95
}

for name, marks in students.items():
    if marks < 35:
        print(name, "- Failed, skipping...")
        continue

    if marks >= 90:
        print(name, "- Excellent! Loop stopped.")
        break

    print(name, "- Passed with", marks, "marks")
```
